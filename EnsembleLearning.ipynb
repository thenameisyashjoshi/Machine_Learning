```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
X = df.drop('species',axis=1)
y = df['species']
```

```python
label = LabelEncoder()
y_enc = label.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_enc, test_size=0.20, random_state=42, stratify=y)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_Knn = KNeighborsClassifier(n_neighbors= 13)
```

```python
model_Knn.fit(X_train,y_train)
```

```python
model_Knn.score(X_test,y_test)
```

```python
from sklearn.svm import SVC
```

```python
model_Svm = SVC(gamma='auto')
```

```python
model_Svm.fit(X_train,y_train)
```

```python
model_Svm.score(X_test,y_test)
```

Now lets use the **GridSearchCV**

GridSearchCV is a Scikit-learn technique that systematically tests different combinations of hyperparameters using cross-validation and selects the combination that performs best according to a chosen evaluation metric.

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_Svm),{
    'C' : [1,10,20,30],
    'kernel' : ['linear','rbf']
},cv=5)
```

```python
classifier.fit(X,y)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

Now lets use the **RandomSearchCV**

RandomizedSearchCV is a Scikit-learn technique that randomly samples a specified number of hyperparameter combinations and evaluates them using cross-validation to identify a good-performing configuration.

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifier_r = RandomizedSearchCV((model_Svm),{
    'C' : [1,10,20,30],
    'kernel' : ['rbf','linear']
},n_iter=4,cv=5,return_train_score=False)
```

```python
classifier_r.fit(X,y)
```

```python
results_r = pd.DataFrame(classifier_r.cv_results_)
```

**Ensemble Learning** :              
Ensemble Learning is a machine learning technique in which multiple models work together to make a prediction, often producing better results than an individual model.

Imagine you're trying to predict whether a passenger survived the Titanic disaster.

You train three different models:

Model 1: Logistic Regression predicts that the passenger survived.
Model 2: Decision Tree predicts that the passenger did not survive.
Model 3: Random Forest predicts that the passenger survived.

If you combine their predictions using majority voting, the final prediction is:

Survived.

Two out of three models predicted survival.

**Types of Ensemble Learning** :

Bagging — Bootstrap Aggregating.

Boosting — Sequentially improving predictive performance.

Stacking — Combining predictions using a meta-model.

**Stacking (Stacked Generalization)**
Stacking is an ensemble technique in which the predictions of multiple base models become inputs to a second-level model called a meta-model.

```python
from sklearn.ensemble import StackingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
```

```python
base_models = [
    ('dt', DecisionTreeClassifier(random_state=42)),
    ('svc',SVC(probability=True, kernel='rbf', random_state=42)),
    ('lr', LogisticRegression(max_iter=1000))
]
```

```python
meta_model = LogisticRegression(max_iter=1000)
```

```python
stacking_clf = StackingClassifier(
    estimators=base_models,
    final_estimator=meta_model,
    cv=5
)
```

```python
stacking_clf.fit(X_train,y_train)
```

```python
y_pred = stacking_clf.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

**Bagging (Bootstrap Aggregating)**

Bagging is an ensemble technique in which multiple models are trained independently on different bootstrap samples of the training data, and their predictions are combined.

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rf_model = RandomForestClassifier(
    n_estimators=100,   #number of Trees
    max_depth = None,
    random_state=42,
    max_features="sqrt"
)
```

```python
rf_model.fit(X_train,y_train)
```

```python
y_pred = rf_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

**Boosting**

Boosting is an ensemble technique in which models are trained sequentially, with later models designed to improve the ensemble's performance.

**AdaBoost Adaptive Boosting, Gradient Boosting, XGBoost**

```python
from sklearn.ensemble import AdaBoostClassifier , GradientBoostingClassifier
from xgboost import XGBClassifier
```

```python
ada_model = AdaBoostClassifier(n_estimators=100, learning_rate=1.0, random_state=None)
```

```python
ada_model.fit(X_train,y_train)
```

```python
y_pred = ada_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

```python
gradient_model.fit(X_train,y_train)
```

```python
y_pred = gradient_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
accuracy
```

```python
xg_model = XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, use_label_encoder=False, eval_matric='mlogloss',random_state=42)
```

```python
xg_model.fit(X_train,y_train)
```

```python
y_pred = xg_model.predict(X_test)
```
