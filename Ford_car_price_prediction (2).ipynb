```python
X = df.drop('price',axis=1)
y = df['price']
```

```python
X
```

```python
X_onehot_encoded = pd.get_dummies(X , columns=['model','transmission','fuelType'])
X_onehot_encoded
```

```python
X_onehot_encoded = X_onehot_encoded.astype(int)
X_onehot_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
columns = ['model','transmission','fuelType']
X_label = X.copy()
for i in columns:
  le = LabelEncoder()
  X_label[i] = le.fit_transform(X_label[i].astype(str))
X_label
```

X_label[i] = le.fit_transform(X_label[i].astype(str))                                                                
we are using .astype(string) is to avoid problems caused by mixed data types or unexpected values.

For example, suppose a column accidentally contains:
petrol,
diesel,
123,
NaN

```python
df['model'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numeric_col = ['year','mileage','tax','mpg','engineSize']
X_onehot_encoded[numeric_col] = scaler.fit_transform(X_onehot_encoded[numeric_col])
X_onehot_encoded
```

```python
X_label[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(X_label[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])

X_label
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

**One-Hot Encoding Prediction**

```python
X_train , X_test , y_train , y_test = train_test_split(X_onehot_encoded,y,test_size=0.2,random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
r2 = r2_score(y_test , y_pred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adj_r2 = 1 -((1-r2)*(n-1)) / (n-p-1)
adj_r2
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
plt.scatter(y_test, y_pred)
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted Price")
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()]
)

plt.show()
```

```python
residuals = y_test - y_pred
plt.scatter(y_pred, residuals)

plt.axhline(0, linestyle='--')

plt.xlabel("Predicted Price")
plt.ylabel("Residual")
plt.title("Residual Plot")

plt.show()
```

```python
train_pred = model.predict(X_train)

train_r2 = r2_score(y_train, train_pred)
test_r2 = r2_score(y_test, y_pred)

print("Train R²:", train_r2)
print("Test R² :", test_r2)
```
